```python
#day-54
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dropout,Dense,LSTM
import matplotlib.pyplot as plt
data=np.sin(np.arange(0,400,0.5))+np.random.randn(800)*0.1
time_steps=60
x,y=[],[]
for i in range(time_steps,len(data)):
    x.append(data[i-time_steps:i])
    y.append(data[i])
x,y=np.array(x),np.array(y)
x=x.reshape(x.shape[0],x.shape[1],1)
print(f" x shape:{x.shape}-this is point")
model=Sequential([
    LSTM(50,return_sequences=True,input_shape=(60,1)),
    Dropout(0.2),
    LSTM(50),
    Dropout(0.2),
    Dense(1)
])
model.compile(optimizer="adam",loss="mse")
model.summary()
history=model.fit(x,y,epochs=10,batch_size=32,validation_split=0.2)
plt.plot(history.history["loss"],label="train")
plt.plot(history.history["val_loss"],label="val")
plt.legend()
plt.title("LSTM")
plt.show()
last_60=data[-60:].reshape(1,60,1)
next_pred=model.predict(last_60)
print(f"next predicted value :{next_pred[0][0]:.4f}")
```

```python
#day-55
import yfinance as yf
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
try:
    df = yf.download('AAPL', period='2y', interval='1d')
    if len(df) < 100:
        df = yf.download('TCS.NS', period='2y', interval='1d')
except:
    df = None

if df is not None and len(df) > 100:
    data = df['Close'].values.reshape(-1,1)
    print(f"Real data downloaded: {len(data)} days")
else:
    data = (np.sin(np.arange(0,400,0.5)) + 150 + np.random.randn(800)*2).reshape(-1,1)
    print("Using dummy stock data (no internet)")
scaler = MinMaxScaler(feature_range=(0,1))
scaled_data = scaler.fit_transform(data)
train_size = int(len(scaled_data)*0.8)
train_data = scaled_data[:train_size]
test_data = scaled_data[train_size:]

print(f"Train: {len(train_data)}, Test: {len(test_data)}")
def create_sequences(data, time_steps=60):
    X, y = [], []
    for i in range(time_steps, len(data)):
        X.append(data[i-time_steps:i, 0])
        y.append(data[i, 0])
    return np.array(X), np.array(y)

time_steps = 60
X_train, y_train = create_sequences(train_data, time_steps)
X_test, y_test = create_sequences(test_data, time_steps)

X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], 1)
X_test = X_test.reshape(X_test.shape[0], X_test.shape[1], 1)

print(f"X_train shape: {X_train.shape}")
model = Sequential([
    LSTM(50, return_sequences=True, input_shape=(time_steps,1)),
    Dropout(0.2),
    LSTM(50),
    Dropout(0.2),
    Dense(1)
])
model.compile(optimizer='adam', loss='mse')
model.summary()
history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))
predicted = model.predict(X_test)
predicted_price = scaler.inverse_transform(predicted)
real_price = scaler.inverse_transform(y_test.reshape(-1,1))
rmse = np.sqrt(np.mean((predicted_price - real_price)**2))
print(f"RMSE: {rmse:.2f}")

plt.figure(figsize=(12,4))
plt.plot(real_price, label='Real Price', color='blue')
plt.plot(predicted_price, label='Predicted Price', color='red')
plt.title(f"Day-55 Real Stock - RMSE: {rmse:.2f}")
plt.legend()
plt.show()

last_60 = scaled_data[-60:].reshape(1,60,1)
tomorrow_scaled = model.predict(last_60)
tomorrow_price = scaler.inverse_transform(tomorrow_scaled)
print(f"Tomorrow predicted price: ${tomorrow_price[0][0]:.2f}")

model.save('stock_lstm.h5')
print("Saved as stock_lstm.h5 - Done!")
```

```python
#day-56

import yfinance as yf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping


df = yf.download('AAPL', period='2y')
df['MA50'] = df['Close'].rolling(50).mean()
df['MA200'] = df['Close'].rolling(200).mean()
df['Returns'] = df['Close'].pct_change()
df.dropna(inplace=True)

features = ['Close', 'MA50', 'Volume']
data = df[features].values
print(f"Data shape: {data.shape} - 502 days x 3 features")

scaler = MinMaxScaler((0,1))
scaled_data = scaler.fit_transform(data)

def create_multistep_seq(data, time_steps=60, future_days=7):
    X, y = [], []
    for i in range(time_steps, len(data)-future_days+1):
        X.append(data[i-time_steps:i]) 
        y.append(data[i:i+future_days, 0])
    return np.array(X), np.array(y)

TIME_STEPS = 60
FUTURE = 7

X, y = create_multistep_seq(scaled_data, TIME_STEPS, FUTURE)
print(f"X: {X.shape}, y: {y.shape}") 

split = int(len(X)*0.8)
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

model = Sequential([
   
    Bidirectional(LSTM(50, return_sequences=True), input_shape=(TIME_STEPS, 3)),
    Dropout(0.2),
    LSTM(50, return_sequences=True),
    Dropout(0.2),
    LSTM(50),
    Dropout(0.2),
    Dense(7) 
])

model.compile(optimizer='adam', loss='mse')
model.summary()

early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

history = model.fit(X_train, y_train, epochs=30, batch_size=32,
                    validation_data=(X_test, y_test),
                    callbacks=[early_stop])


def inverse_close(scaled_close_seq):

    dummy = np.zeros((len(scaled_close_seq), 3))
    dummy[:,0] = scaled_close_seq[:,0] # take first day for demo
    return scaler.inverse_transform(dummy)[:,0]


pred_scaled = model.predict(X_test)

close_scaler = MinMaxScaler()
close_scaler.min_, close_scaler.scale_ = scaler.min_[0], scaler.scale_[0]

pred_7days = []
real_7days = []
for i in range(len(pred_scaled)):
    pred_7days.append(close_scaler.inverse_transform(pred_scaled[i].reshape(-1,1)).flatten())
    real_7days.append(close_scaler.inverse_transform(y_test[i].reshape(-1,1)).flatten())

pred_7days = np.array(pred_7days)
real_7days = np.array(real_7days)


plt.figure(figsize=(14,6))
plt.plot(real_7days[0], label='Real Next 7 Days', marker='o', color='blue', linewidth=2)
plt.plot(pred_7days[0], label='Predicted Next 7 Days', marker='x', color='red', linewidth=2)

plt.fill_between(range(7), pred_7days[0]*0.95, pred_7days[0]*1.05, color='red', alpha=0.2, label='Confidence Band')
plt.title("Day-56 Bi-LSTM: Next 7 Days Prediction with Confidence Band")
plt.xlabel("Future Day (1-7)")
plt.ylabel("Price $")
plt.legend()
plt.grid(True)
plt.show()


last_close = float(df['Close'].values[-1])
tomorrow_pred = pred_7days[-1][0]
print(f"Last Close: ${last_close:.2f}")
print(f"Tomorrow Pred: ${tomorrow_pred:.2f}")
print(f"Signal: {'BUY 📈' if tomorrow_pred > last_close else 'SELL 📉'}")

last_60 = scaled_data[-60:].reshape(1,60,3)
next_7_scaled = model.predict(last_60)
next_7_price = close_scaler.inverse_transform(next_7_scaled[0].reshape(-1,1)).flatten()

print("\nNext 7 Days Forecast:")
for i, price in enumerate(next_7_price, 1):
    print(f"Day {i}: ${price:.2f}")

model.save('bilstm_7day.h5')
print("Saved bilstm_7day.h5")
```
